This section offers a basic overview and demonstration of how to manipulate CSV files in Python.

## Functions Used in this Section

This section is desiged to provide an overview and review of some of the
basic functions that are useful to work with CSV data in Python.

- Working with [CSV in Python](https://docs.python.org/3/library/csv.html) 
  - `.reader()` and `.DictReader()` - these functions are used to read in, or "load" data. Both can be used to read any CSV file, but the former requires you to reference the data using indices, while the latter allows for referring to data elements by name, as you might from a Python dictionary. The latter dictionary approach requires that you know the field names in the CSV or have a CSV which includes field names in a header row.  
  - `.writer()` and `.DictWriter()` - these functions are used to write out, or save data. The latter is useful if you can refer to data values with a known key name (as in a dictionary), but requires more knowledge of the data structure you want to write.

In most cases, this book will use the dictionary functions, because it is easier to refer to known data elements by name.

```python
import csv
```

First, you should take a look at the file. So, open it in another tab,
or if you are working on a bash system, you can use `!cat`.

Assuming this is a file that represents field observations of wildlife during a
given time period in a given spot (we don't know much about the context beyond this),
then we can use the basic `.reader()` function to create a more 'readable' output:

```python
with open('../data/animals.csv', encoding='utf-8') as f:
    data = csv.reader(f, delimiter=',')
    line_count = 0
    for row in data:
        line_count += 1 
        print(f'On {row[0]} there was a count of {row[2]} reported {row[1]}.')
    print(f'\nRead {line_count} lines.')
```

## Using the DictReader functions

If you know or can apply header names for each of the columns,
you can treat these as a list and the `csv` library will create
a dictionary. This is useful if you want to refer to values
by their names (not by column index).

```python
cols = ['date','animal','count']

with open('../data/animals.csv', encoding='utf-8') as f:
    data = csv.DictReader(f, fieldnames=cols, delimiter=',')
    line_count = 0
    for row in data:
        line_count += 1 
        print(row)
#        print(f'On { row['date'] } there was a count of {row['count']} reported {row['animal']}.')
    print(f'\nRead {line_count} lines.')
```

```python
cols = ['date','animal','count']

with open('../data/animals.csv', encoding='utf-8') as f:
    data = csv.DictReader(f, fieldnames=cols, delimiter=',')
    line_count = 0
    for row in data:
        line_count += 1 
        date = row['date']
        animal = row['animal']
        count = row['count']
        print(f'On { date } there was a count of { count } reported { animal }.')
    print(f'\nRead {line_count} lines.')
```
