This section offers a basic overview and demonstration of how to manipulate CSV files in Python.

## Functions Used in this Section

This section reviews some of the
basic functions that are useful to know about when working with CSV data in Python.

- Working with [CSV in Python](https://docs.python.org/3/library/csv.html) 
  - `.reader()` and `.DictReader()` - these functions are used to read in, or "load" data. Both can be used to read any CSV file, but the former requires you to reference the data using indices, while the latter allows for referring to data elements by name, as you might from a Python dictionary. The latter dictionary approach requires that you know the field names in the CSV or have a CSV which includes field names in a header row. As illustrated below, with a CSV that lacks a header row, the header must be supplied as a list. 
  - `.writer()` and `.DictWriter()` - these functions are used to write out, or save data. The latter is useful if you can refer to data values with a known key name (as in a dictionary), but requires more knowledge of the data structure you want to write.

In most cases, this book will use the dictionary functions, because it is easier to refer to known data elements by name.

### Reading CSV into Python

To begin, use the basic `.reader()` function to make the data accessible to Python.
Although the demonstration CSV should process consistently with relatively few additional instructions, there are a few useful conventions to note. Within the `with open()` construction, it is good practice to specify the text encoding with the `encoding` argument (shown below with UTF-8, which is a good default to choose when possible), and providing the `newline` argument with a blank string will help to process the data correctly, recognizing only newline indicators at the end of each data record. Finally, in the `.reader()` function below, note the specification of `delimiter`. By default, this will be a comma, but other delimiters are possible (see [](#csv-delimiters)).

```python
with open('../data/animals.csv', encoding='utf-8', newline='') as f:
    data = csv.reader(f, delimiter=',')
    row_count = 0
    for row in data:
        row_count += 1 
        print(f'On {row[0]} there was a count of {row[2]} reported {row[1]}.')
    print(f'\nRead {row_count} rows.')
```

#### Using the DictReader functions

If you know or can apply header names for each of the columns,
you supply these a list, which is passed to the `.DictReader()`'s `fieldnames` argument, then Python will be able to process this data like 
a dictionary.
This is useful if you want to refer to values
by their names (not by column index).

:::{hint} Different Delimiters
:label: csv-delimiters
CSVs typically use a comma as a delmiter, and indeed that is part of the file format name ("Comma Separated Values").
Even so, other delimiters are frequently found. Sometimes, this is reflected in the name, such as TSV for "Tab Separated Values."
Sometimes, however, a different delimiter may be found even in a file with the `.csv` extension. Common delimiters include the pipe (`|`), semicolon, and tab (notated in Python as `\t`), among others. 

Determine the delimiter with a quick file inspection, and provide it in the `delimiter` argument of the reader, if necessary. (If you skip the argument, the library assumes it is a comma.)
:::

```python
cols = ['date','animal','count']

with open('../data/animals.csv', encoding='utf-8') as f:
    data = csv.DictReader(f, fieldnames=cols, delimiter=',')
    row_count = 0
    for row in data:
        row_count += 1 
        print(row)
    print(f'\nRead {row_count} rows.')
```

```python
cols = ['date','animal','count']

with open('../data/animals.csv', encoding='utf-8') as f:
    data = csv.DictReader(f, fieldnames=cols, delimiter=',')
    row_count = 0
    for row in data:
        row_count += 1 
        date = row['date']
        animal = row['animal']
        count = row['count']
        print(f'On { date } there was a count of { count } reported { animal }.')
    print(f'\nRead {row_count} lines.')
```

### Writing Data with CSV

For the most part, this is a basic operation. It may be useful to note a few good practices
for writing more reusable CSVs. In addition to the `.writer()` and `.DictWriter()` functions, it may be useful to know about the `.writeheader()` function, which will write a header row (if this is supported by your data),
as well as the specification of text encoding, using the `encoding=` argument in the file write code. Following the header, the `.writerows()` function completes the output by writing each data record as a new row in the resulting CSV. (Note there is a corresponding `.writerow()` function that can be used to write individually, if required by your incoming data.)

Below is an example of writing out the data using `.DictWriter()`:

```python
# if the cols variable was not initiated previously, uncomment below:
#cols = ['date','animal','count']

fname = '../output/animals-rewrite.csv'

with open('../data/animals.csv', 'r', encoding='utf-8', newline='') as f:
    rows = list(csv.DictReader(f, fieldnames=cols))

with open(fname, 'w', encoding='utf-8', newline='') as f_out:
    writer = csv.DictWriter(f_out, fieldnames=cols)
    writer.writeheader()
    writer.writerows(rows)

print(f'Wrote {fname}')
```

The above rewrites the CSV and illustrates how to write out a new CSV with a header row and ensures it is encoded as UTF-8 text. If you compare the new output file with the original, you will now see the header row on the first line.
