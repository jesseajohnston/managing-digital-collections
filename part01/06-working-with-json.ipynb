This section reviews some of the basics on working with JSON data using Python tools.
The main data manipulation functions use Python's standard `json` library.
In many ways, JSON can be managed like a dictionary in Python, and its plain-text data format,
consisting of curly braces and square brackets mimics the notation used by Python
to represent dictionaries.

## Functions Used in this Section

This section is desiged to provide an overview and review of some of the
main uses and functions for JSON data in Python.

- Working with [JSON in Python](https://docs.python.org/3/library/json.html) 
  - `.load()` and `.loads()` - these functions are used to read in, or "load" data. The former function is used for loading from a file, while the latter is used to load string formatted data.
  - `.dump()` and `.dumps()` - these functions are used to write out, or "dump" data. The former function is useful for outputting a binary or file type object, while the latter is mostly used to write out strings.

### Reading JSON into Python

To convert JSON into a processable object for python, use the JSON module and the `.loads()` function. For example:

```python
import json
```

```python
stringOfJsonData = '{ "Results": 2, "isPrinted" : true, "format" : null, "books": [{ "title": "Gone with the Wind", "author":"Mitchell" },{ "title": "Wuthering Heights", "author": "Bronte" } ]}'

jsonDataAsPythonValue = json.loads(stringOfJsonData)

jsonDataAsPythonValue
```

```python
for key in jsonDataAsPythonValue:
    print(key)
```

Print the keys & values

```python
for key in jsonDataAsPythonValue:
    print(key, ':', jsonDataAsPythonValue[key])
```

```python
for book in jsonDataAsPythonValue['books']:
    print(book)
```

Another possibility is to assign the data element you want to a variable (e.g., `books` below).
Then, you can continue to work with that data on its own:

```python
books = jsonDataAsPythonValue['books']

for book in books:
    print(book)
```

### Writing JSON from Python

To write out the data into a JSON file, use the `.dumps()` function. For example:

```python
jsonDataAsPythonValue
```

```python
# uncomment next line if you haven't yet imported the json module
#import json

stringOfJsonData = json.dumps(jsonDataAsPythonValue)

stringOfJsonData
```

```python
# write to a file

with open('../data/json-books.json', 'w', encoding='utf-8') as f:
    f.write(json.dumps(jsonDataAsPythonValue, indent=2))
    print('wrote',f.name)
```
