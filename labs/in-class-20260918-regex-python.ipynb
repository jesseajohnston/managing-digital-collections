```python
subfield_regex = re.compile(r"\<subfield ")
```

```python
subfield_matches = re.findall(subfield_regex, xml_test_string)

print(f"Regex findall returned: {type(subfield_matches)} {len(subfield_matches)}")
```

```python
for match in subfield_matches:
    print(match.span())
```

```python
# working to match dates

iso_date_regex = re.compile(r"\d{4}-\d{2}-\d{2}")

for date_match in re.finditer(iso_date_regex, xml_test_string):
    print(date_match)
    #print(date_match.start)
    #print(xml_test_string[int(date_match.start):int(date_match.end)])
```

```python
# working with groups
www_regex = re.compile(r'https?:\/\/www\..*')

for www_match in re.finditer(www_regex, xml_test_string):
    print(www_match)
```

```python
# working with groups
www_regex_groups = re.compile(r'https?:\/\/(\w{3,4})\..*')

for www_match in re.finditer(www_regex_groups, xml_test_string):
    print(www_match)
    print('group 1 (subdomain):',www_match.group(1))
```
