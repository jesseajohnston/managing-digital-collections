```python
# either provide or give these from another source, name of the file you want to write, column headers 
collection_info_csv = OUT / 'collection_items_data_foromekaimport.csv'
headers = ['id','date','title','contributors']
items_written = 0 # counter to monitor how many items were found

# delete existing file if it already exists
if collection_info_csv.is_file():
    collection_info_csv.unlink()

# now, loop through all the files
for item_file in list_of_item_metadata_files:
    print('opening',item_file)
    with open(item_file, 'r', encoding='utf-8') as item:
        item_data = json.load(item)
    #    for element in item_data.keys():
    #        print(element, ':', item_data[element])
        try:
            item_title = item_data['title']
        except:
            item_title = 'No title found'
        item_id = item_data['id']
        try:
            item_date = item_data['date']
        except:
            item_date = 'No date'
        try:
            item_contributors = item_data['contributor_names']
            # for contributor names, you'll need to have multiple options, that can handle a string or an array, and that will be more complex than try/except
        except:
            item_contributors = 'None listed'

        row_dict = dict()

        row_dict['id'] = item_id
        row_dict['date'] = item_date
        row_dict['title'] = item_title
        row_dict['contributors'] = item_contributors[0]

        # write to the csv
        with open(collection_info_csv, 'a', encoding='utf-8') as fout:
            writer = csv.DictWriter(fout, fieldnames=headers)
            if items_written == 0:
                writer.writeheader()
                print('wrote header')
            writer.writerow(row_dict)
            items_written += 1

            print('adding data for',item_id)

print('wrote the list file',collection_info_csv)
print(f'\nTransformed item data for {str(items_written)} items.')
```
