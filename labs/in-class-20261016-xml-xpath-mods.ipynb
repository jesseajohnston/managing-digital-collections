```python
import re
from pathlib import Path
```

```python
# import the data
DATA = Path('..', 'data')
OUT = Path('..', 'output')
```

```python
MODS_collection_5 = DATA / 'lcwa-mods-5.xml'

tree = etree.parse(MODS_collection_5)
root = tree.getroot()

print(root[:50])
```

```python
c = 0 
for element in root.iter():
    c += 1
    print(element.tag)

print(c)
```

```python
# establish namespace dictionary
ns = {
    'mods': 'http://www.loc.gov/mods/v3'
}
```

```python
MODS_collection_25 = DATA / 'lcwa-mods-25.xml'

xml_records_25 = etree.parse(MODS_collection_25)
metadata = xml_records_25.getroot()

for subject in metadata.findall('.//mods:subject', namespaces=ns):
    print(subject.tag, ' -- ', subject.attrib)
#    print(etree.tostring(subject))
```

```python
for subject in metadata.findall('.//mods:subject[@authority="lcsh"]', namespaces=ns):
    print(subject.tag, ' -- ', subject.attrib)
#    print(etree.tostring(subject))
```

```python
for subject in metadata.findall('.//mods:subject[@authority="lcsh"]', ns):
    subject.set('displayLabel', 'Blank')
    subject.set('updated', 'true')
    print(subject.tag, ' -- ', subject.attrib)
```

```python
## save the file

newfile = OUT / 'mods_test_update.xml'

# writing the data
metadata.write(newfile, xml_declaration=True, encoding='utf-8', pretty_print=True)
print('Tried to write the file! :-)')
```
