Demonstrations of basic XML parsing. Follows the general steps in <https://jesseajohnston.github.io/managing-digital-collections/part02/xml-01-basics-with-et/>. Next week we discuss namespaces and working with XML data.

```python
import xml.etree.ElementTree as ET
```

```python
from pathlib import Path

DATA = Path('..','data')

EAD_sample = DATA / 'book-sample-in-class.xml'

#EAD_relpath = 'data/book-sample-in-class.xml'
```

```python
print(EAD_sample)
```

```python
# parse XML = load + process

tree = ET.parse(EAD_sample)
root = tree.getroot()
```

```python
print(root)
```

```python
print(ET.tostring(root))
```

```python
for element in root:
    print(element.tag,' --- ',element.attrib, ' --- ',element.text)
```

```python
for element in root.iter():
    print(element.tag, ' --- ', element.attrib)
```

```python
for element in root.iter():
    print(element.tag, ':', element.text)
```

```python
book = tree.find('{http://www.nogood.com/Book}book')
print(book.tag, book.attrib, book.text)
```

```python
for element in book:
    print(element.tag, element.text)
```

```python
books = root.findall('.//{http://www.nogood.com/Book}book')
print(books)
```
