```python
import requests
import json
import csv
```

```python
# item IDs from lab 5
item_ids = ['cph.3f05183', 'fsa.8d24709', 'highsm.64003']
```

```python
# set up an API call info
baseURL = 'https://www.loc.gov'
endpoint = '/resource/'

params = {
    'fo':'json'
}
```

```python
r = requests.get(baseURL + endpoint + item_ids[0], params=params)
```

```python
r.url
```

## Now, use this to get the JSON response & save it!

```python
r.json()
```

```python
r.json().keys()
```

```python
item_metadata = r.json()['item']
```

```python
item_metadata
```

```python
item_metadata.keys()
```

```python
for k in item_metadata.keys():
    print(k, ':', item_metadata[k])
```

# Next: retrieve multiple items

:tada:

```python
item_ids
```

```python
for id in item_ids:
    r = requests.get(baseURL + endpoint + id, params=params)
    print(r.url)
    print(r.status_code)
    item_data = r.json()['item']
    with open(id + '_metadata.json', 'w', encoding='utf-8') as f:
        f.write(json.dumps(item_data, indent=2))
```
